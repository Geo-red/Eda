```python
import pandas as pd
import numpy as np
hairfall = [50,40,39,20,54,72,69,67,31,15,0,9,87]
bins = [10,20,30,40,50]
category = pd.cut(hairfall, bins)
category
```

```python
pd.value_counts(category)
```

```python
 pd.value_counts(category)
```

```python
bin_names = ['No Hairfall','Normal Hairfall','Livable hairfall','Soon to be bald']
pd.cut(hairfall, bins, labels=bin_names)
```
