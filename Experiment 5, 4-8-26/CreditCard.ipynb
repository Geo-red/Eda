```python
import pandas as pd
fraud_test_df = pd.read_csv('/content/fraudTest.csv')
print("Fraud Test DataFrame head:")
display(fraud_test_df.head())
```

```python
data_for_cut = fraud_test_df['amt']
import pandas as pd
bins = [0, 50, 200, 1000, data_for_cut.max() + 1]
category = pd.cut(data_for_cut, bins)
print(category)
```

```python
print(pd.Series(category).value_counts())
```

```python
bin_names = ['Small Amount', 'Medium Amount', 'Large Amount', 'Very Large Amount']
print(pd.cut(data_for_cut, bins, labels=bin_names))
```

```python
import numpy as np

data_for_equal_bins = fraud_test_df['city_pop']
print(pd.cut(data_for_equal_bins, 5, precision=0))
```
