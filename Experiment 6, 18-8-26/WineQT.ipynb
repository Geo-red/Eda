```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("WineQT.csv")
print(df.head(3))
print(df.tail(3))
```

```python
df.dtypes
```

```python
print(df.describe())
pd.isna(df).any()
```

```python
mean = df["alcohol"].mean()
median =df["alcohol"].median()
mode = df["alcohol"].mode()
print(mean , median, mode)
```

```python
df.quality.value_counts().nlargest(10).plot(kind='bar', figsize=(10,6))
plt.title("Number of wines by quality")
plt.ylabel('Number of wines')
plt.xlabel('Wine Quality')
```

```python
sns.FacetGrid(df).map(sns.histplot,"alcohol").add_legend()
```

```python
sns.FacetGrid(df).map(sns.histplot,"quality").add_legend()
```

```python
sns.boxplot(x="quality",data=df)
plt.show()
```

```python
plt.scatter(df["alcohol"], df["quality"])
plt.title("Scatter Plot for Alcohol vs Quality")
plt.xlabel("Alcohol")
plt.ylabel("Quality")
```

```python
sns.pairplot(df,vars = ['quality', 'alcohol','pH'], kind="reg")
plt.show()
```

```python
sns.set(style="ticks", color_codes=True)
sns.pairplot(df,height=2,vars = ['fixed acidity', 'alcohol','pH'], hue="quality")
plt.show()
```

```python
from scipy import stats
corr = stats.pearsonr(df["quality"], df["alcohol"])
print("p-value:\t", corr[1])
print("cor:\t\t", corr[0])
```

```python
correlation = df.corr(method='pearson', numeric_only=True)
correlation
```

```python
sns.heatmap(correlation,xticklabels=correlation.columns,
 yticklabels=correlation.columns)
```
